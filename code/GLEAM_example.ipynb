# GLEAM: run the imputation

Python kernel from the **conda environment of GLEAM** (needs pandas, numpy, scipy, scikit-learn, torch, torchvision, torch_geometric, torch_sparse). To make the environment available in Jupyter:

```bash
conda activate <gleam_env>
python -m ipykernel install --user --name gleam
```

and pick the kernel `gleam` for this notebook. Needed in the working directory: `GLEAM_impute.py`.

Input: the files written by `SAVI_prepare_gleam_inputs()` (see `SAVI_data_prep_example.ipynb`) in `work_dir`.
Output (also in `work_dir`): `sp_rna_exist_withatac_all.csv`, `sp_rna_pred_withatac_all.csv`, `sp_atac_exist_withatac.csv`.

````python
import os
import GLEAM_impute

work_dir = "/storage10/shuyang/MISAR-seq-main/More_evaluated_gene/"   # same folder as in the R notebook
gpu = 0                                              # GPU index; None = current device
````

````python
# check that the input files from the R step are there
needed = ["sc_rna_raw.csv", "sp_rna_raw.csv", "sc_atac_scaled.csv", "sp_atac_scaled.csv",
          "sc_cca_emb.csv", "sp_cca_emb.csv", "sc_cca_knn.csv", "sp_cca_knn.csv",
          "sc_rna_atac_knn.csv", "sp_rna_atac_knn.csv", "sp_spatial_knn.csv"]
missing = [f for f in needed if not os.path.exists(os.path.join(work_dir, f))]
print("missing input files:", missing)
assert not missing, "run SAVI_prepare_gleam_inputs() in R first"
````

````python
# train GLEAM and write the imputation files (takes a while)
# options: epochs=600, hidden_channels=140, lr=0.001, seed=0, skip_atac=False
GLEAM_impute.run(work_dir,hidden_channels=100, gpu=gpu)
````

````python
# the three files that SAVI_load_imputation() reads in R
for f in ["sp_rna_exist_withatac_all.csv", "sp_rna_pred_withatac_all.csv", "sp_atac_exist_withatac.csv"]:
    print(f, os.path.exists(os.path.join(work_dir, f)))
````
